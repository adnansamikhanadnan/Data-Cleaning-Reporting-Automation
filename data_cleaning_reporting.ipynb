# Data Cleaning & Reporting Automation
Automate cleaning of messy employee data and generate a report and visual summary.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 1. Load raw data

```python
data = pd.read_csv('raw_employee_data.csv')
data.head()
```

## 2. Inspect missing values and duplicates

```python
print('Rows:', len(data))
print('Duplicate rows:', data.duplicated().sum())
print('\nMissing values:')
print(data.isna().sum())
```

## 3. Clean the data

```python
data = data.drop_duplicates().copy()
for col in ['Name','Email','City','Department']:
    data[col] = data[col].astype('string').str.strip()
data['Name'] = data['Name'].str.title()
data['City'] = data['City'].str.title()
data['Department'] = data['Department'].str.title()
data['Salary'] = pd.to_numeric(data['Salary'], errors='coerce')
data['Salary'] = data['Salary'].fillna(data['Salary'].median())
data['Email'] = data['Email'].fillna('not_provided@example.com')
data['JoinDate'] = pd.to_datetime(data['JoinDate'], errors='coerce')
data['JoinDate'] = data['JoinDate'].fillna(pd.Timestamp('2026-01-01'))
data['JoinDate'] = data['JoinDate'].dt.strftime('%Y-%m-%d')
data.head()
```

## 4. Save cleaned data

```python
data.to_csv('cleaned_employee_data.csv', index=False)
print('Saved cleaned_employee_data.csv')
```

## 5. Generate department summary

```python
summary = data.groupby('Department').agg(EmployeeCount=('EmployeeID','count'), AverageSalary=('Salary','mean'), TotalSalary=('Salary','sum')).reset_index()
summary['AverageSalary'] = summary['AverageSalary'].round(2)
summary
```

## 6. Create visual report

```python
plt.figure(figsize=(9,6))
plt.bar(summary['Department'], summary['AverageSalary'])
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.title('Average Salary by Department')
plt.tight_layout()
plt.savefig('department_summary.png', dpi=150)
plt.show()
```

## Conclusion
The workflow automatically cleans inconsistent data, saves a clean dataset, summarizes departments, and produces a visual report.
